```python
def categorise_col(df,col,labels):
  edges=[df[col].describe()['min'],
         df[col].describe()['25%'],
         df[col].describe()['50%'],
         df[col].describe()['75%'],
         df[col].describe()['max']]

  df[col]=pd.cut(df[col],edges,labels=labels,duplicates='drop')
  return df
```

```python
labels=['not_popular','below_avg','average','popular']
```

```python
categorise_col(df,'Vote_Average',labels)
```

```python
df['Vote_Average'].value_counts()
```

```python
df.dropna(inplace=True)

df.isna().sum()
```

#we'd split generes into a list andthen explode our dataframe to have only one gener per roe for each movie

```python
# Step 1: Convert all values to string and split by comma
df['Genre'] = df['Genre'].astype(str).str.split(',')

# Step 2: Explode the Genre list into rows
df = df.explode('Genre').reset_index(drop=True)

# Step 3: Clean up leading/trailing spaces
df['Genre'] = df['Genre'].str.strip()

# Step 4: Optional — replace 'nan' strings with real NaN
df['Genre'] = df['Genre'].replace('nan', pd.NA)

# Step 5: Drop rows where Genre is empty or NaN (if needed)
df = df.dropna(subset=['Genre'])


```

```python
#Casting Columns into Category
df['Genre']=df['Genre'].astype('category')
df['Genre'].dtypes
```
